```python
import pandas as pd
from pathlib import Path
from sqlalchemy import create_engine

engine = create_engine("postgresql+psycopg2://olist:olist@localhost:5432/olist")
OUT = Path("../data/export"); OUT.mkdir(parents=True, exist_ok=True)

for t in ["core.dim_date", "core.dim_customer", "core.dim_product", "core.dim_seller",
          "core.fact_orders", "core.fact_order_items",
          "marts.seller_performance", "marts.rfm"]:
    df = pd.read_sql(f"SELECT * FROM {t}", engine)
    df.to_csv(OUT / f"{t.split('.')[1]}.csv", index=False)
    print(f"{t:30s} {len(df):>8d} linhas")
```

```python
pd.read_sql("""SELECT ROUND(SUM(price)::numeric, 2) AS receita,
                      COUNT(DISTINCT order_id)       AS pedidos
               FROM core.fact_order_items""", engine)
```
